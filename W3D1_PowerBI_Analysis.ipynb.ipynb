# Week 3 Day 1 – Power BI

## Project: Indian Business EDA

### Objective
To create a Power BI report using cleaned Indian business data and visualize the distribution of businesses across cities.

### Dataset
W2D4_Cleaned_Business_Data.csv

### Tool Used
Microsoft Power BI Desktop

## 1. Business Distribution by City

Created a clustered column chart in Power BI to understand the number of businesses in each city.

### Results

| City | Number of Businesses |
|---|---:|
| Bangalore | 2 |
| Chennai | 1 |
| Hyderabad | 1 |
| Mumbai | 1 |

### Visualization
- Chart Type: Clustered Column Chart
- X-axis: City
- Y-axis: Count of Business_ID
- Data Labels: Enabled
- Title: Number of Businesses by City

## 2. Interactive City Slicer

Created an interactive slicer in Power BI using the City field.

### Features
- Added City as the slicer field.
- Displayed Bangalore, Chennai, Hyderabad and Mumbai.
- Selecting a city filters the business distribution chart.
- Helps users explore business data city-wise.

## 3. Key Findings

- Bangalore has the highest number of businesses (2).
- Chennai, Hyderabad and Mumbai have 1 business each.
- The dataset contains 5 businesses across 4 cities.

## 4. Conclusion

Power BI helped transform the cleaned business dataset into an interactive visual report. The column chart and city slicer make it easier to understand and explore business distribution.
