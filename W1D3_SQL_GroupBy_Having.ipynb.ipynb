```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("indian_business.db")

print("Database connected successfully!")
```

```python
query = "SELECT * FROM businesses"
df = pd.read_sql_query(query, conn)

df
```

```python
query = """
SELECT City, COUNT(*) AS Total_Businesses
FROM businesses
GROUP BY City;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT 
    City,
    COUNT(*) AS Total_Businesses,
    AVG(Business_ID) AS Average_ID,
    SUM(Business_ID) AS Total_ID
FROM businesses
GROUP BY City;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT City, COUNT(*) AS Total_Businesses
FROM businesses
GROUP BY City
HAVING COUNT(*) > 1;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT City, COUNT(*) AS Total_Businesses
FROM businesses
WHERE Business_ID >= 102
GROUP BY City;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT City, COUNT(*) AS Total_Businesses
FROM businesses
GROUP BY City
ORDER BY Total_Businesses DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    City,
    MIN(Business_ID) AS Minimum_ID,
    MAX(Business_ID) AS Maximum_ID
FROM businesses
GROUP BY City;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT City, COUNT(*) AS Total_Businesses
FROM businesses
GROUP BY City
HAVING COUNT(*) >= 2;
"""

result = pd.read_sql_query(query, conn)
result
```

## SQL concepts practiced
- GROUP BY
- COUNT()
- SUM()
- AVG()
- HAVING
- WHERE with GROUP BY
- ORDER BY with GROUP BY
- MIN()
- MAX()

## Findings
- GROUP BY groups rows based on common column values.
- COUNT() counts records in each group.
- SUM() calculates the total of values in a group.
- AVG() calculates the average of values in a group.
- MIN() finds the smallest value.
- MAX() finds the largest value.
- HAVING filters grouped results.
- WHERE filters rows before grouping.
- ORDER BY sorts the grouped results.
