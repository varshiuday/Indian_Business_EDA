```python
import sqlite3
import pandas as pd
import numpy as np

# Connect to database
conn = sqlite3.connect("indian_business.db")

print("Database connected successfully!")

# Display available tables
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

print("\nAvailable tables:")
print(tables)
```

```python
df = pd.read_sql_query(
    "SELECT * FROM businesses",
    conn
)

print("Dataset loaded successfully!")
print("\nDataset shape:", df.shape)

display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Convert Business Name values to uppercase
df["Business_Name_Upper"] = df["Business Name"].apply(
    lambda x: x.upper()
)

print("Updated dataset:")
display(df)
```

```python
# Display the actual column names
print("Columns:", df.columns.tolist())

# Use the second column (Business Name)
df["Business_Name_Upper"] = df.iloc[:, 1].apply(
    lambda x: str(x).upper()
)

print("\nUpdated dataset:")
display(df)
```

```python
# Check for duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Find businesses whose names contain "Cafe"
cafe_businesses = df[
    df.iloc[:, 1].astype(str).str.contains(
        "Cafe", case=False, na=False
    )
]

print("Businesses containing 'Cafe':")
display(cafe_businesses)
```

```python
# Count businesses in each city
city_summary = df.groupby(df.columns[2]).size().reset_index(name="Business_Count")

# Save the summary as a CSV file
city_summary.to_csv("city_summary.csv", index=False)

print("City summary exported successfully!")
display(city_summary)
```
