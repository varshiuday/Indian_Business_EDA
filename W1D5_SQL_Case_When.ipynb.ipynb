# W1D5 — SQL CASE WHEN & Conditional Analysis

## Objective
Learn how to use CASE WHEN statements to categorize and analyze data conditionally.

## Concepts
- CASE WHEN
- ELSE
- END
- Conditional categorization
- Multiple conditions
- CASE WHEN with ORDER BY

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("indian_business.db")

print("Database connected successfully!")
```

```python
query = "SELECT * FROM businesses"

df = pd.read_sql_query(query, conn)
df
```

```python
query = """
SELECT
    Business_ID,
    Business_Name,
    CASE
        WHEN Business_ID >= 104 THEN 'High ID'
        ELSE 'Low ID'
    END AS ID_Category
FROM businesses;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    Business_ID,
    Business_Name,
    CASE
        WHEN Business_ID <= 102 THEN 'Beginner'
        WHEN Business_ID <= 104 THEN 'Intermediate'
        ELSE 'Advanced'
    END AS Business_Level
FROM businesses;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    Business_Name,
    City,
    CASE
        WHEN City = 'Bangalore' THEN 'Major Hub'
        WHEN City = 'Mumbai' THEN 'Major Hub'
        ELSE 'Other City'
    END AS City_Category
FROM businesses;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Business_ID, Business_Name, City
FROM businesses
ORDER BY
    CASE
        WHEN City = 'Bangalore' THEN 1
        WHEN City = 'Mumbai' THEN 2
        ELSE 3
    END;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    Business_ID,
    Business_Name,
    City,
    CASE
        WHEN Business_ID < 103 THEN 'Category A'
        WHEN Business_ID < 105 THEN 'Category B'
        ELSE 'Category C'
    END AS Business_Category
FROM businesses;
"""

result = pd.read_sql_query(query, conn)
result
```

## Objective

## SQL concepts practiced
- CASE WHEN
- ELSE
- END
- Multiple conditions
- Conditional categorization
- CASE WHEN with ORDER BY

## Findings
- CASE WHEN creates conditional logic inside SQL queries.
- WHEN specifies a condition.
- THEN returns a result when the condition is satisfied.
- ELSE provides a default result when no condition matches.
- END terminates the CASE expression.
- Multiple WHEN conditions can classify data into different categories.
- CASE WHEN can be used with ORDER BY for custom sorting.
