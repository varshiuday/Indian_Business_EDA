```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("indian_business.db")

print("Database connected successfully!")
```

```python
query = "SELECT * FROM businesses"

df = pd.read_sql_query(query, conn)

df
```

```python
query = """
SELECT Business_ID, Business_Name, City
FROM businesses
WHERE Business_ID > (SELECT AVG(Business_ID) FROM businesses);
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Business_ID, Business_Name, City
FROM businesses
WHERE City IN (
    SELECT City
    FROM businesses
    GROUP BY City
    HAVING COUNT(*) > 1
);
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Business_ID, Business_Name, City
FROM businesses
WHERE Business_ID > (
    SELECT Business_ID
    FROM businesses
    WHERE Business_Name = 'GreenLeaf'
);
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT City, Total_Businesses
FROM (
    SELECT City, COUNT(*) AS Total_Businesses
    FROM businesses
    GROUP BY City
) AS City_Counts
ORDER BY Total_Businesses DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Business_ID, Business_Name, City
FROM businesses
WHERE Business_ID < (
    SELECT AVG(Business_ID)
    FROM businesses
);
"""

result = pd.read_sql_query(query, conn)
result
```

## Findings
- A subquery is a query inside another SQL query.
- Subqueries can be used inside WHERE and FROM clauses.
- AVG() can be used inside a subquery to calculate average values.
- IN can check whether a value matches values returned by a subquery.
- Comparison operators can compare values against subquery results.
- Subqueries help filter and retrieve data dynamically.
