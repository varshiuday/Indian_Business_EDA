# W1D2 — SQL Joins

## Objective
Learn how SQL JOINs combine data from multiple tables.

## Concepts
- INNER JOIN
- LEFT JOIN
- RIGHT JOIN
- FULL OUTER JOIN

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("indian_business.db")

businesses = pd.DataFrame({
    "Business_ID": [101, 102, 103, 104, 105],
    "Business_Name": ["FreshMart", "TechZone", "GreenLeaf", "Cafe Aroma", "StyleHub"],
    "City": ["Bangalore", "Mumbai", "Chennai", "Bangalore", "Hyderabad"]
})

owners = pd.DataFrame({
    "Business_ID": [101, 102, 103, 106],
    "Owner_Name": ["Rahul", "Priya", "Arjun", "Sneha"]
})

businesses.to_sql("businesses", conn, if_exists="replace", index=False)
owners.to_sql("owners", conn, if_exists="replace", index=False)

print("Businesses and Owners tables created successfully!")
```

```python
query = """
SELECT businesses.Business_Name, owners.Owner_Name
FROM businesses
INNER JOIN owners
ON businesses.Business_ID = owners.Business_ID;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT businesses.Business_Name, owners.Owner_Name
FROM businesses
LEFT JOIN owners
ON businesses.Business_ID = owners.Business_ID;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT businesses.Business_Name, owners.Owner_Name
FROM businesses
RIGHT JOIN owners
ON businesses.Business_ID = owners.Business_ID;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT businesses.Business_Name, owners.Owner_Name
FROM businesses
FULL OUTER JOIN owners
ON businesses.Business_ID = owners.Business_ID;
"""

result = pd.read_sql_query(query, conn)
result
```

# W1D2 — SQL Joins

## Objective
Learn how SQL JOINs combine data from multiple tables.

## Tables created
- businesses
- owners

## JOIN concepts practiced
- INNER JOIN: Returns matching records from both tables.
- LEFT JOIN: Returns all businesses and matching owners.
- RIGHT JOIN: Returns all owners and matching businesses.
- FULL OUTER JOIN: Returns all records from both tables, matching where possible.

## Findings
- JOINs connect tables using a common column such as Business_ID.
- INNER JOIN returns only matching records.
- LEFT JOIN keeps all records from the left table.
- RIGHT JOIN keeps all records from the right table.
- FULL OUTER JOIN keeps matching and unmatched records from both tables.
- Missing matches appear as NULL or NaN.
