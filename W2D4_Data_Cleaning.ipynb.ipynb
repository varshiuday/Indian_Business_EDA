```python
import pandas as pd
import numpy as np
import sqlite3

print("Libraries imported successfully!")
```

```python
conn = sqlite3.connect("indian_business.db")

print("Database connected successfully!")
```

```python
df = pd.read_sql_query("SELECT * FROM businesses", conn)

print("Original dataset:")
display(df)
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
df.iloc[:, 1] = df.iloc[:, 1].fillna("Unknown")
df.iloc[:, 2] = df.iloc[:, 2].fillna("Unknown")

print("Missing values handled!")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed successfully!")
print("Remaining rows:", len(df))
```

```python
df.iloc[:, 1] = df.iloc[:, 1].astype(str).str.strip()
df.iloc[:, 2] = df.iloc[:, 2].astype(str).str.strip()

print("Text cleaning completed!")
display(df)
```

```python
df.iloc[:, 2] = df.iloc[:, 2].str.title()

print("City names standardized:")
print(df.iloc[:, 2])
```

```python
print("Data types before conversion:")
print(df.dtypes)
```

```python
df.iloc[:, 0] = pd.to_numeric(
    df.iloc[:, 0],
    errors="coerce"
).astype("Int64")

print("Data types after conversion:")
print(df.dtypes)
```

```python
print("Invalid or missing Business IDs:")
print(df[df.iloc[:, 0].isna()])
```

```python
print("Cleaned dataset:")
display(df)

print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```
