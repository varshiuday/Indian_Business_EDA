# Week 3 Day 2 – DAX Basics

## Project: Indian Business EDA

### Objective
To understand DAX measures in Power BI and calculate business metrics.

### DAX Measures Created

1. Total Businesses
   - COUNT(Business_ID)
   - Expected result: 5

2. Total Cities
   - DISTINCTCOUNT(City)
   - Expected result: 4

### Key Learning
DAX measures calculate values dynamically and respond to report filters such as the City slicer.

## Practical Implementation

### 1. Total Businesses

DAX Formula:

Total Businesses = COUNT('W2D4_Cleaned_Business_Data'[Business_ID])

Result: 5 businesses

### 2. Total Cities

DAX Formula:

Total Cities = DISTINCTCOUNT('W2D4_Cleaned_Business_Data'[City])

Result: 4 cities

## Key Learnings

- COUNT() counts non-blank values.
- DISTINCTCOUNT() counts unique values.
- Measures dynamically respond to report filters.
- The City slicer changes the Total Businesses calculation.

## Conclusion

DAX measures help calculate important business metrics and make Power BI reports interactive.
