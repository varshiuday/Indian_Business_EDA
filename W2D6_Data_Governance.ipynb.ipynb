```python
import pandas as pd
import numpy as np
import sqlite3
import json

print("Libraries imported successfully!")
```

```python
conn = sqlite3.connect("indian_business.db")

print("Database connected successfully!")
```

```python
df = pd.read_sql_query("SELECT * FROM businesses", conn)

print("Dataset loaded successfully!")
display(df)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
business_id_col = df.columns[0]
business_name_col = df.columns[1]
city_col = df.columns[2]

print("Data quality rules:")
print("1. Business ID must not be missing.")
print("2. Business ID must be greater than zero.")
print("3. Business Name must not be empty.")
print("4. City must not be empty.")
print("5. Business ID must be unique.")
```

```python
data_contract = {
    "dataset_name": "Indian Business Dataset",
    "required_columns": [
        business_id_col,
        business_name_col,
        city_col
    ],
    "rules": {
        "business_id_not_null": True,
        "business_id_positive": True,
        "business_id_unique": True,
        "business_name_not_empty": True,
        "city_not_empty": True
    }
}

print("Data contract created:")
print(json.dumps(data_contract, indent=4))
```

```python
quality_checks = {
    "Business ID not null": df[business_id_col].notna().all(),
    "Business ID positive": (df[business_id_col] > 0).all(),
    "Business ID unique": df[business_id_col].is_unique,
    "Business Name not empty": (
        df[business_name_col].notna()
        & df[business_name_col].astype(str).str.strip().ne("")
    ).all(),
    "City not empty": (
        df[city_col].notna()
        & df[city_col].astype(str).str.strip().ne("")
    ).all()
}

for rule, passed in quality_checks.items():
    print(rule, ":", "PASS" if passed else "FAIL")
```

```python
total_checks = len(quality_checks)
passed_checks = sum(quality_checks.values())

quality_score = (passed_checks / total_checks) * 100

print("Total checks:", total_checks)
print("Passed checks:", passed_checks)
print("Data quality score:", quality_score, "%")
```

```python
quality_report = pd.DataFrame({
    "Rule": list(quality_checks.keys()),
    "Status": [
        "PASS" if result else "FAIL"
        for result in quality_checks.values()
    ]
})

display(quality_report)
```

```python
with open("W2D6_Data_Contract.json", "w") as file:
    json.dump(data_contract, file, indent=4)

print("Data contract saved successfully!")
```

```python
quality_report.to_csv(
    "W2D6_Data_Quality_Report.csv",
    index=False
)

print("Data quality report exported successfully!")
```

```python
import sqlite3
import pandas as pd

# Connect to the project database
conn = sqlite3.connect("indian_business.db")

# Load business data
df = pd.read_sql_query("SELECT * FROM businesses", conn)

conn.close()

# Display the dataset
print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
# Create Great Expectations context
context = gx.get_context()

# Create a Pandas Data Source
data_source = context.data_sources.add_pandas(
    name="business_data_source"
)

# Create a DataFrame Data Asset
data_asset = data_source.add_dataframe_asset(
    name="business_data_asset"
)

# Create a Batch Definition
batch_definition = data_asset.add_batch_definition_whole_dataframe(
    name="business_batch"
)

# Build a batch using our DataFrame
batch = batch_definition.get_batch(
    batch_parameters={"dataframe": df}
)

print("Great Expectations DataFrame connection successful!")
print("Batch created:", batch)
```

```python
# Create an Expectation Suite
suite = gx.ExpectationSuite(
    name="business_data_quality"
)

# Define data quality rules
expectations = [
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="Business ID"
    ),
    gx.expectations.ExpectColumnValuesToBeUnique(
        column="Business ID"
    ),
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="Business ID",
        min_value=1
    ),
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="Business Name"
    ),
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="City"
    )
]

# Add expectations to the suite
for expectation in expectations:
    suite.add_expectation(expectation)

# Save the suite in the context
suite = context.suites.add(suite)

print("Data quality expectations created successfully!")
print("Total expectations:", len(suite.expectations))
```

```python
# Create a Validator
validator = context.get_validator(
    batch=batch,
    expectation_suite=suite
)

# Run validation
validation_results = validator.validate()

# Display validation results
print("Business Data Validation Results")
print("--------------------------------")

for result in validation_results.results:
    print(
        result.expectation_config.type,
        ":", 
        "PASSED" if result.success else "FAILED"
    )

print("--------------------------------")
print("Overall success:", validation_results.success)
```

```python
# Create a new expectation suite
suite = gx.ExpectationSuite(
    name="business_data_quality_corrected"
)

# Define expectations using actual column names
expectations = [
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="Business_ID"
    ),
    gx.expectations.ExpectColumnValuesToBeUnique(
        column="Business_ID"
    ),
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="Business_ID",
        min_value=1
    ),
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="Business_Name"
    ),
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="City"
    )
]

# Add expectations
for expectation in expectations:
    suite.add_expectation(expectation)

# Save the corrected suite
suite = context.suites.add(suite)

print("Corrected expectation suite created!")
print("Total rules:", len(suite.expectations))
```

```python
# Create validator with corrected suite
validator = context.get_validator(
    batch=batch,
    expectation_suite=suite
)

# Run validation
validation_results = validator.validate()

# Display results
print("Corrected Business Data Validation")
print("----------------------------------")

for result in validation_results.results:
    status = "PASSED" if result.success else "FAILED"
    print(result.expectation_config.type, ":", status)

print("----------------------------------")
print("Overall success:", validation_results.success)
```

```python
# Count passed and failed expectations
total_checks = len(validation_results.results)

passed_checks = sum(
    result.success for result in validation_results.results
)

failed_checks = total_checks - passed_checks

# Calculate quality score
quality_score = (
    passed_checks / total_checks * 100
    if total_checks > 0 else 0
)

# Create report
quality_report = pd.DataFrame({
    "Metric": [
        "Total Checks",
        "Passed Checks",
        "Failed Checks",
        "Quality Score (%)",
        "Overall Status"
    ],
    "Value": [
        total_checks,
        passed_checks,
        failed_checks,
        round(quality_score, 2),
        "PASSED" if failed_checks == 0 else "NEEDS IMPROVEMENT"
    ]
})

display(quality_report)
```

```python
quality_report.to_csv(
    "W2D6_Data_Quality_Report.csv",
    index=False
)

print("Data quality report saved successfully!")
```
