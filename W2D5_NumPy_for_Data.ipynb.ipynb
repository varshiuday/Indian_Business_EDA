```python
import numpy as np
import pandas as pd
import sqlite3

print("Libraries imported successfully!")
```

```python
business_ids = np.array([101, 102, 103, 104, 105])

print("Business IDs:")
print(business_ids)
```

```python
print("Array dimensions:", business_ids.ndim)
print("Array shape:", business_ids.shape)
print("Array size:", business_ids.size)
print("Array data type:", business_ids.dtype)
```

```python
print("Addition:", business_ids + 10)
print("Subtraction:", business_ids - 10)
print("Multiplication:", business_ids * 2)
print("Division:", business_ids / 2)
```

```python
print("Mean:", np.mean(business_ids))
print("Median:", np.median(business_ids))
print("Minimum:", np.min(business_ids))
print("Maximum:", np.max(business_ids))
print("Standard deviation:", np.std(business_ids))
```

```python
large_ids = business_ids[business_ids > 102]

print("Business IDs greater than 102:")
print(large_ids)
```

```python
conn = sqlite3.connect("indian_business.db")

df = pd.read_sql_query("SELECT * FROM businesses", conn)

print("Business dataset loaded successfully!")
display(df.head())
```

```python
business_ids = df.iloc[:, 0].to_numpy()

print("NumPy Business IDs:")
print(business_ids)
```

```python
print("Average Business ID:", np.mean(business_ids))
print("Highest Business ID:", np.max(business_ids))
print("Lowest Business ID:", np.min(business_ids))
print("Total Business IDs:", np.sum(business_ids))
```

```python
summary = {
    "Total Businesses": len(business_ids),
    "Average ID": np.mean(business_ids),
    "Minimum ID": np.min(business_ids),
    "Maximum ID": np.max(business_ids)
}

print("NumPy Summary:")
for key, value in summary.items():
    print(key, ":", value)
```

```python
print("========== W2D5 FINAL CHECK ==========")

print("NumPy version:", np.__version__)
print("Dataset rows:", df.shape[0])
print("Array size:", business_ids.size)
print("Average ID:", np.mean(business_ids))

print("=====================================")
```

```python
numpy_summary = pd.DataFrame([summary])

numpy_summary.to_csv(
    "W2D5_NumPy_Summary.csv",
    index=False
)

print("NumPy summary exported successfully!")
display(numpy_summary)
```
